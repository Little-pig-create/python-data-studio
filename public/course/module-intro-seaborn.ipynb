## 本章场景



用 Matplotlib 画图很好，但每次都要写一堆设置配色、误差线、分组的代码。当你想"比较几个类别的销售额差异""看两个变量的相关趋势"时，更希望一句话出图。



**Seaborn** 建立在 Matplotlib 之上，专为**统计图表**设计：给它一张**整洁的长表**（一行一个观测、一列一个字段），它就能自动处理分组、配色、误差线等统计细节，一行代码得到一张专业图表。它是"探索数据时快速画统计图"的最佳选择。

- **迁移**：能对一组分组数据画出统计图并比较不同分组的差异。


## 44.1 这个模块解决什么问题

**背景引入**：有时你想快速查看"不同类型商品的平均销量差多少、数据分布如何"，逐步手写 Matplotlib 太慢。

打个比方：Matplotlib 像一盒需要自己动手拼的积木——颜色、刻度、图例都得你一块块搭；Seaborn 则像一份说明书完备的'统计图积木包'——你只要把整洁的长表、x/y/hue 报给它，它就把分组、配色、误差线这些重复劳动一手包办。想快速'看一眼数据什么样'，选它最省力。

- **长表即输入**：给 data + x/y/hue 就能出图，不用手动分组；

- **统计自动化**：自动含误差线、回归线、密度曲线等统计细节；

- **多变量对比**：用 hue / col / row 一次看多个分组维度。

## 44.3 第一个上手示例



感受"一行代码出统计图"。运行下方代码：画出不同品类销量的分布箱线图。


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

plt.rcParams["font.sans-serif"] = ["font.sans-serif", "Microsoft YaHei", "SimHei", "DejaVu Sans"]
plt.rcParams["axes.unicode_minus"] = False


orders = pd.DataFrame(
    {
        "品类": ["餐饮"] * 3 + ["购物"] * 3 + ["交通"] * 3,
        "金额": [35.5, 48, 52, 120, 299, 150, 18, 25, 30],
    }
)
sns.boxplot(data=orders, x="品类", y="金额")
plt.title("不同品类金额分布")
plt.show()
```
