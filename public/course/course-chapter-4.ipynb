## 4.1 创建、索引与切片



**背景引入**：账目不是一笔，而是成百上千笔。把多笔金额、多条记录放在一个容器里才能统一处理——列表就是 Python 组织"多条数据"的第一种容器。（把它想成一排带编号的储物格：按格子号存取，还记着放的先后顺序。）先掌握怎么创建、怎么按位置取数据。



### 4.1.1 创建



列表用方括号创建，元素用逗号分隔：amounts = [35.5, 18.0, 299.0, 12.0]。列表**可变**——可以原地增删元素，这与字符串不同。



### 4.1.2 索引与切片



- 索引：amounts[0] 第一笔，amounts[-1] 最后一笔；

- 切片：amounts[:2] 前两笔；

- 长度：len(amounts) 返回元素个数。



### 4.1.3 步长切片（跳着取）



**背景引入**：有时想"隔着取"——比如只统计每隔一笔的金额、或把最近的账目倒序看一遍。切片还能带**步长**：



`amounts[::1]` 全部；`amounts[::2]` 每两个取一个；`amounts[::-1]` 倒序（从后往前）。



- 语法：`列表[起点:终点:步长]`，步长为负表示方向反转；

- `[::-1]` 是最常用的"一键倒序"写法，不会修改原列表；

- 步长 1 是默认，正步长从头走、负步长从尾走。



> **口诀**：要逆序找 [::-1]，要跳着取找 [::n]。





```python
# 步长切片：每隔一笔 / 倒序查看

amounts = [35.5, 18.0, 299.0, 12.0, 88.0]

print("全部:", amounts)

print("隔一个取:", amounts[::2])  # 每次跳过 1 个

print("倒序:", amounts[::-1])  # 从后往前

print("倒序隔一个:", amounts[::-2])
```

## 4.6 易错点提醒



### 4.6.1 append、extend 与 insert 别混用

- append 把参数作为**一个整体**追加；
- extend 把参数中的每个元素依次追加；
- insert(索引, 元素) 在指定位置插入。

用 append 追加列表会把整个列表嵌套进去——这是常见错误。



```python
categories = ["餐饮", "交通"]
categories.append("购物")
categories.extend(["医疗", "其他"])
categories.insert(1, "工资")
print(categories)

wrong = ["餐饮"]
wrong.append(["交通", "购物"])
print(wrong)
```

**输出解读**：append 加一个元素，extend 展开追加，insert 在索引 1 插入；而 append 一个列表会得到嵌套列表。


### 4.6.2 remove、pop、del 与 clear 怎么选

| 需求 | 方法 |
| --- | --- |
| 按值删除第一个匹配 | remove(元素) |
| 删除并拿到返回值 | pop(索引) |
| 删除一段（切片） | del 列表[:n] |
| 清空全部 | clear() |



```python
categories = ["餐饮", "交通", "购物", "其他"]
categories.remove("交通")
removed = categories.pop(1)
del categories[:1]
print("pop 返回：", removed)
print("剩余：", categories)
categories.clear()
print("清空后：", categories)
```

**输出解读**：remove 按值删除；pop 返回被删元素；del 支持切片删除；clear 清空后列表仍可继续使用。


### 4.6.3 index、count 与成员检查

in 检查是否存在，count 统计次数，index 返回第一次出现的位置。先用 in 判断再取 index，避免 ValueError。



```python
categories = ["餐饮", "交通", "餐饮", "购物"]
print("餐饮" in categories)
print(categories.count("餐饮"))
if "购物" in categories:
    print("购物第一次在索引：", categories.index("购物"))
if "医疗" not in categories:
    print("医疗还没有出现")
```

**输出解读**：成员检查返回 True；count 统计出现 2 次；index 返回 3；不存在的元素用 in 提前判断。


### 4.6.4 sort 与 sorted：一个改原列表，一个不改

sort() 原地排序（修改原列表）；sorted() 返回新列表（原列表不变）。对字典列表必须用 key 指定排序依据。



```python
records = [
    {"date": "2026-08-07", "amount": 18.0},
    {"date": "2026-08-06", "amount": 35.5},
]
by_date = sorted(records, key=lambda item: item["date"])
by_amount = sorted(
    records,
    key=lambda item: item["amount"],
    reverse=True,
)
print("原顺序：", records)
print("按日期：", by_date)
print("按金额降序：", by_amount)
```

**输出解读**：sorted 不改变原顺序；按日期与按金额降序得到两个不同的排序结果。


### 4.6.5 copy 与切片都是浅复制

copy() 与切片只复制**外层容器**；如果列表元素本身是可变对象（如字典、列表），内层对象仍共享。修改内层会影响原列表。



```python
records = [{"category": "餐饮", "tags": ["工作餐"]}]
shallow = records.copy()
shallow.append({"category": "交通", "tags": []})
print("外层长度：", len(records), len(shallow))
shallow[0]["tags"].append("可报销")
print("原列表内部：", records[0])
print("复制列表内部：", shallow[0])
```

**输出解读**：外层长度不同（浅复制不影响外层增删），但内层 tags 被共享——浅复制列表中的字典仍是同一个对象。

## 教学实验：边界条件

这一组实验专门训练“观察一个结果 → 只改一个变量 → 解释变化”。先运行第一个代码单元格，再运行第二个。


```python
orders = [
    {"order_id": "A01", "amount": 280},
    {"order_id": "A02", "amount": 300},
    {"order_id": "A03", "amount": 520},
]
threshold = 300
for order in orders:
    label = "达到门槛" if order["amount"] >= threshold else "未达到门槛"
    print(order["order_id"], order["amount"], label)
```

### 第一个结果怎么读

这里的重点不是记住 `if`，而是观察 `>=` 如何处理刚好等于 300 的记录。边界条件必须和业务规则保持一致。

请记录：输入是什么、输出是什么、输出支持了哪一个结论。


```python
threshold = 500
for order in orders:
    label = "重点关注" if order["amount"] >= threshold else "普通订单"
    print(order["order_id"], "->", label)
print("重点订单数：", sum(order["amount"] >= threshold for order in orders))
```

### 第二个结果怎么读

只把门槛从 300 改成 500，再比较标签和数量变化。这个实验训练的是“改一个输入，解释一个输出”。

## 错误恢复：类型转换失败怎么办

真实数据和真实代码都会出问题。本节先观察问题，再用一个明确的检查或修复步骤恢复运行。


```python
amount_text = "128.5"
try:
    amount = int(amount_text)
except ValueError as error:
    print("第一次转换失败：", type(error).__name__)
    amount = float(amount_text)
print("可以继续使用的金额：", amount)
```

### 错误恢复步骤

1. 先看错误类型、字段或数据形状。
2. 判断问题发生在输入、处理中间结果还是输出。
3. 修复后重新检查结果，而不是只让代码不报错。

先读错误类型，再决定修复方法。这里不是盲目忽略错误，而是明确知道整数转换不适合带小数的文本。
