## 0. 认识你的工作台

Notebook 由一个个**单元格（cell）**组成，只有两种需要关心的类型：

| 类型 | 长什么样 | 用途 |
| --- | --- | --- |
| 代码单元格 | 左侧有 `In []` 标记 | 写 Python 代码，按运行键执行 |
| Markdown 单元格 | 没有输入标记 | 写说明文字，比如你现在读的这一段 |

**操作约定**：本章所有代码单元格都要亲手运行一次。运行方式是选中单元格后
按 `Shift + Enter`（或点工具栏的运行按钮）。


## 1.1 把 Python 当计算器：print 与四则运算

**概念**：`print()` 是最常用的函数，作用是"把括号里的内容显示出来"。
Python 里加 `+`、减 `-`、乘 `*`、除 `/` 直接写，和数学一样。


### 例 1｜最小例子：打印一句话，算一笔乘法


````python
print("你好，记账助手")

# 一支笔 3.5 元，买 4 支，一共多少钱？
print(3.5 * 4)
````

**输出解读**：两行结果。第一行是引号里的文字原样显示；第二行是计算结果 `14.0`。
注意：只有写在 `print()` 括号里的内容才会显示，"运行"和"显示"是两件事。


### 例 2｜业务例子：一笔外卖账单


````python
# 一单外卖：主食 25 元，配送费 4 元，用了一张满 20 减 5 的券
print("餐费:", 25)
print("配送费:", 4)
print("优惠:", 5)
print("实付:", 25 + 4 - 5)
````

**输出解读**：`print` 括号里可以用逗号隔开多项，输出时中间自动加一个空格。
实付 24 元就是这四个数按业务规则算出来的结果。

> 场景切换说明：本章例子都围绕"个人记账"。这一主线会贯穿整个 Python 基础模块，
> 数据从一笔账逐步长成一个月的账本。


### 例 3｜常见错误：以为写出来就会显示

初学者最常见的第一个困惑："我明明写了算式，为什么没输出？"
**原因**：没套 `print()`。在 Notebook 里，一个单元格**默认只显示最后一行**的表达式结果，
上面的行什么都不显示。


````python
# 反例：下面第一行的结果会"消失"，只有最后一行被显示出来
25 + 4 - 5  # 算了，但没人看见（不显示）
3.5 * 4  # 恰好是最后一行，所以只有它显示
````

````python
# 正确做法：想显示的都交给 print
print(25 + 4 - 5)
print(3.5 * 4)
````

**要点**：拿不准时，凡是想看到的结果就写 `print()`。这不丢人，
专业代码里 `print` 也是排查问题的第一工具。


## 1.2 注释：写给人看的说明

**概念**：以 `#` 开头的内容是**注释**，Python 会直接忽略，不参与运行。
注释是写给自己和同事看的，用来解释"这段代码为什么这么写"。

```python
# 这是注释，运行时被忽略
print("这行会执行")  # 行尾也可以写注释
```


### 例 1｜最小例子：注释的三种位置


````python
# 整行注释：说明下面这段代码的目的
price = 3.5  # 行尾注释：说明这个变量的含义
# print("这行被注释掉了，不会执行")
print(price)
````

**输出解读**：只输出 `3.5`。整行注释和被注释掉的 print 都没有执行。

### 例 2｜业务例子：用注释交代口径

记账代码里最怕"数字来历不明"。每个关键数字都应有一句注释说明口径。


````python
# 2026-08 账单汇总
breakfast = 6 * 30  # 早餐 6 元/天，按 30 天计
coffee = 18 * 12  # 咖啡 18 元/杯，每周 3 杯约 12 杯/月
print("固定开销合计:", breakfast + coffee)
````

**输出解读**：`198`。三个月后回看这段代码，你仍然能从注释知道 6、30、18、12 各是什么——
这就是注释的价值。

### 例 3｜常见错误：用中文标点写注释或代码

初学者高频错误：在**代码区**使用了中文的引号 `“”` 或括号 `（）`。
Python 只认英文标点；中文标点会直接导致语法错误（`SyntaxError`）。


````python
# 反例：中文引号会让 Python 无法理解
# print("你好")   <- 注意这里的引号是中文引号，取消注释运行会报 SyntaxError

# 正确做法：代码与注释里的标点都用英文半角
print("你好")
````

**怎么发现这类错误**：报错信息 `SyntaxError` 通常会把出错位置用 `^` 指出来。
看到 `SyntaxError` 先检查：引号、括号、逗号是不是不小心打成了中文输入法。


## 1.3 变量与运行顺序：Notebook 的"记忆"

**概念**：上一章预告过，变量是给数据起的名字。这里要先讲清 Notebook 特有的一点：

> **单元格的执行顺序由你点击的顺序决定，而不是从上到下的书写顺序。**

运行过的代码会被 Python"记住"（存在内核里）。这就是为什么有时你会遇到
`NameError`（名字错误）——使用了还没定义的变量。


### 例 1｜最小例子：先定义，再使用


````python
month = "2026-08"
total = 1024.5
print(month, "共支出", total, "元")
````

**输出解读**：`2026-08 共支出 1024.5 元`。这个单元格运行后，`month` 和 `total`
就被内核记住了，**后面的单元格可以直接用**。

### 例 2｜业务例子：跨单元格累计

真实的记账是分步进行的：先记各笔账，再在后面的单元格里汇总。


````python
# 第 1 步：记录 8 月第一周的账
week1 = 156.5
week2 = 189.0
print("前两周合计:", week1 + week2)
````

````python
# 第 2 步：使用上一个单元格的变量
week3 = 143.5
week4 = 201.0
month_total = week1 + week2 + week3 + week4
print("8 月总支出:", month_total)
````

**输出解读**：第二个单元格能直接用 `week1`、`week2`，是因为它们在**更早运行的**单元格中
已定义。如果跳过第一个单元格直接运行第二个，就会看到 `NameError`。

### 例 3｜常见错误：乱序运行导致变量"张冠李戴"

下面演示一个隐蔽陷阱：同一个名字被赋值两次，结果取决于**最后一次运行**。


````python
# 反例：如果先运行本格，再运行上面的例 2 格，month_total 会变成错误组合
week1 = 0.0  # 假设你不小心把这格里的 week1 清零后重新运行
print("当前 week1 =", week1)
````

**要点**：
- 变量的值以**最后一次运行**的赋值为准；
- 迷茫时的标准恢复动作：**重启内核（Kernel → Restart），然后从上到下重新运行**；
- 写笔记、改代码时尽量保持"从上到下运行一遍全部通过"的状态。


## 1.4 报错不是失败：读懂第一类错误信息

**概念**：报错信息的固定结构是——最后一行是**错误类型**，上面是**出错位置**。
初学者只需要先学会两件事：找到最后一行的类型名、往上找到指向自己代码的那一行。


### 例 1｜最小例子：故意触发的 NameError（看，不跑）


````python
# 反例：取消下一行注释并运行，观察报错结构
# print(monthly_total)   # NameError: name 'monthly_total' is not defined

# 解读（运行上面后回来读）：
# 1. 最后一行 NameError 是错误类型：用了不存在的名字；
# 2. is not defined 表示"未定义"——通常是拼写错误或没运行定义它的单元格。
# 修复：检查拼写，或先运行定义该变量的单元格。
print("本格演示了如何解读 NameError，但本身可以安全运行")
````

**为什么要"注释着放"反例**：如果直接把报错代码留在格子里运行，
整个笔记从上到下就"跑不通"了。所以本课程约定：
**反例代码默认注释掉，想观察报错时手动取消注释运行一次，再把注释加回去。**

### 例 2｜业务例子：三种新手最常见的错误对照


````python
# 错误 1：NameError —— 名字没定义（拼写或没运行）
# print(month_totall)        # 多打了一个 l

# 错误 2：TypeError —— 类型不匹配（文字和数字不能直接相加）
# print("总支出: " + 1024.5) # TypeError；应写 print("总支出: " + str(1024.5))

# 错误 3：SyntaxError —— 语法错误（少了引号/括号，或用了中文标点）
# print("总支出: 1024.5)     # 少了右引号

print("三种错误的示例都被注释保护，取消对应行注释即可观察")
````

**输出解读**：遇到报错的排查顺序——
1. 看最后一行的错误类型（NameError / TypeError / SyntaxError …）；
2. 找到指向你代码的行号，回到那一行；
3. 对照上面的三类典型原因检查；
4. 改完从报错的单元格开始重新运行。

## 易错点清单

- 忘写 `print()`，以为代码没执行；
- 代码里混入中文标点（引号、括号、逗号），报 `SyntaxError`；
- 单元格乱序运行，变量值和书写的直觉不一致；
- 用了没定义（或拼错）的名字，报 `NameError`；
- `print` 写成 `Print`——Python 区分大小写。
