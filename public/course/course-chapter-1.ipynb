## 1. 先认识计算机：程序运行在哪里

计算机可以看成一个“接收输入、执行指令、保存数据、产生输出”的系统。学习 Python 前，先建立以下基础概念：

| 概念 | 作用 | 在 Python 学习中的对应物 |
| --- | --- | --- |
| CPU | 执行指令和计算 | 执行 Python 解释器发出的操作 |
| 内存（RAM） | 临时保存正在使用的数据 | 保存变量、列表、DataFrame 等对象 |
| 存储设备 | 长期保存文件 | 保存 `.py`、`.ipynb`、`.csv` 和图片 |
| 操作系统 | 管理硬件、文件和程序 | 提供路径、进程和运行环境 |
| 程序 | 按规则组织的指令 | Python 源代码和 Notebook 单元格 |
| 输入/输出 | 程序与外部世界交换信息 | 读取文件、接收参数、打印结果、绘制图表 |

可以用一句话概括：**代码是指令，变量是内存中的数据，文件是持久化数据，解释器负责执行代码。**

## 2. 从程序到结果：Python 代码是怎样运行的

Python 是一种高级、通用、解释型编程语言。我们编写的是接近人类表达方式的源代码，Python 解释器负责读取代码、执行运算并返回结果。

一个最小程序通常包含四个环节：

1. **输入**：变量、键盘输入或数据文件；
2. **处理**：计算、判断、循环、函数调用或数据转换；
3. **输出**：文本、表格、图形或文件；
4. **检查**：确认类型、数量、范围和结果是否合理。

Python 适合本课程的原因包括：语法相对简洁、标准库丰富、数据分析生态成熟、可用于自动化和机器学习，并且便于把分析过程记录为可复现文档。它不是“只能做数据分析”的语言，也可以用于 Web、脚本、测试、科学计算和人工智能。

```python
# 这段代码同时展示了：输入、处理、输出和检查
learner = "Python 学习者"
years = 2026 - 2025  # 这是一次简单的数值计算
if not isinstance(learner, str) or years < 0:
    print('输入检查未通过')
else:
    print('输入检查通过')
print("身份:", learner)
print("示例计算结果:", years)
print("变量类型:", type(years).__name__)
```

## 3. 为什么本课程使用 Jupyter Notebook

Jupyter Notebook 不是另一种 Python 语言，而是一种**交互式编程与教学文档环境**。一个 `.ipynb` 文件可以同时保存 Markdown 说明、Python 代码、运行输出、表格、图形和结论。

与只提交一个 `.py` 文件相比，Notebook 更适合本课程的主要原因：

- **边学边运行**：可以只运行当前单元格，立即观察变量和结果；
- **过程可见**：输入、清洗、计算、建模和解释可以按步骤排列；
- **文字与代码结合**：Markdown 可以记录问题、假设、数据口径和结论；
- **适合数据分析**：表格、图片和模型指标可以直接嵌入结果区域；
- **便于复现**：重新启动内核并从上到下运行，可以检查流程是否完整；
- **适合反馈**：教师可以针对某个单元格、输出或图表给出具体建议。

Notebook 也有局限：它容易因为乱序执行产生隐藏状态，不能把“页面上看起来有输出”误认为“从头运行一定成功”。因此本课程始终强调：**重启内核、从上到下运行、保留关键输出、解释结果边界。**

## 4. Notebook 的基本组成和运行机制

| 组成 | 用途 | 学习时的注意点 |
| --- | --- | --- |
| Markdown 单元格 | 标题、说明、问题、结论 | 说明“为什么做”和“结果意味着什么” |
| Code 单元格 | 编写并执行 Python | 一次完成一个清晰步骤 |
| Kernel（内核） | 保存变量并执行代码 | 重启内核会清空变量状态 |
| Output（输出） | 显示文本、表格、图形或错误 | 输出应带标签、单位和口径 |
| 执行编号 | 反映实际运行顺序 | `In [5]` 不一定是页面第 5 个单元格 |

推荐的学习循环是：**阅读说明 → 运行代码 → 查看输出 → 修改一个参数 → 重新运行 → 写下解释**。这比只复制代码更能建立编程和分析能力。

```python
# 观察 Notebook 中“变量状态”的特点
course_name = "Python Data Studio"
chapter_count = 108
print("课程:", course_name)
print("章节数量:", chapter_count)
print("变量存储在当前内核中:", isinstance(chapter_count, int))
```

## 示例 3：组织一次微型分析

按“输入数据 → 指标计算 → 结果输出”组织代码。

```python
scores = [82, 91, 76, 88, 95]
average = sum(scores) / len(scores)
best = max(scores)
print(f"平均分: {average:.1f}")
print(f"最高分: {best}")
```

## 示例 4：检查输入和结果

先检查类型与长度，再输出指标，避免空列表或类型错误被忽略。

```python
scores = [82, 91, 76, 88, 95]
valid_scores = (
    isinstance(scores, list)
    and len(scores) > 0
    and all(isinstance(score, (int, float)) for score in scores)
)
print('成绩输入检查:', '通过' if valid_scores else '未通过')
average = sum(scores) / len(scores)
print("样本数:", len(scores))
print("平均分:", round(average, 1))
```

## 示例 5：验证执行顺序

修改变量后重新运行依赖单元格，比较前后结果，理解内核中的状态变化。

```python
base_hours = 2
students = 5
total_hours = base_hours * students
print("第一次计算:", total_hours)

base_hours = 3
print("只修改变量后:", total_hours)
print("重新计算后:", base_hours * students)
```

## 示例 6：把参数和报告分开

把业务参数放在一个字典里，计算和输出只读取参数，修改输入时不用到处找数字。

```python
plan = {
    "course": "Python数据分析",
    "weeks": 4,
    "hours_per_week": 3,
}
total_hours = plan["weeks"] * plan["hours_per_week"]
report = (
    f"{plan['course']}：{plan['weeks']} 周 × "
    f"每周 {plan['hours_per_week']} 小时 = {total_hours} 小时"
)
print(report)
```

## 示例 7：检查类型、单位和边界

数值正确不等于口径正确；同时检查类型、单位和允许范围。

```python
days = 14
minutes_per_day = 45
hours_per_day = minutes_per_day / 60

valid_plan = (
    isinstance(days, int)
    and isinstance(minutes_per_day, int)
    and days > 0
    and 0 < minutes_per_day <= 24 * 60
)
print('学习计划检查:', '通过' if valid_plan else '未通过')
print("天数类型:", type(days).__name__)
print(f"每天学习: {hours_per_day:.2f} 小时")
```

## 示例 8：受控地观察常见报错

先让错误以可读方式出现，再补上缺失变量；不要用 `try/except` 静默吞掉所有异常。

```python
try:
    print(total_minutes)
except NameError as error:
    print("变量尚未定义:", type(error).__name__)

days = 14
minutes_per_day = 45
total_minutes = days * minutes_per_day
print("补齐变量后，总分钟数:", total_minutes)
```

## 示例 9：记录一次可复核输出

输出中同时保留输入、计算口径和结果，别人才能判断数字是否算对。

```python
inputs = {"订单数": 128, "完成订单数": 96}
completion_rate = inputs["完成订单数"] / inputs["订单数"]
print("输入:", inputs)
print(f"完成率 = 完成订单数 / 订单数 = {completion_rate:.1%}")
if not 0 <= completion_rate <= 1:
    print('完成率检查未通过，请检查订单数和完成订单数')
else:
    print('完成率检查通过')
```

## 常见误区

- 依赖变量尚未创建就运行后续单元格
- 修改变量后未重新运行依赖它的单元格
- 只保留结果，不保留输入和计算过程
- 把分钟、小时、金额等不同单位混在同一个变量里
- 用一个很大的 try/except 把真正的错误隐藏起来
