## 核心概念

- Markdown 单元格保存标题、规则和结论；代码单元格执行 Python 代码。
- `In [n]` 是实际执行顺序，不代表单元格在页面中的位置。
- 变量存活于当前内核；重启内核会清空变量、执行编号和输出状态。
- 后续代码依赖前面已经执行的变量；依赖关系应通过单元格顺序表达。
- 输出是检查计算的证据，不能只依赖最后一个汇总数字。
- 用 `type()`、`len()`、断言和格式化输出检查输入与结果。
- 把输入参数、计算过程和结论分开，后续修改参数时更容易定位差异。
- Notebook 的可复现性来自“相同输入 + 相同顺序 + 相同环境”，不是来自页面看起来完整。


## 单元格怎么分工

| 单元格 | 适合放什么 | 运行后重点检查 |
| --- | --- | --- |
| Markdown | 问题、口径、假设、结论和操作提示 | 读者能否知道这一段要解决什么 |
| 输入代码 | 常量、列表、字典或数据文件路径 | 类型、单位、样本量是否符合预期 |
| 计算代码 | 一种明确的计算或转换 | 中间变量是否有清晰名称 |
| 检查代码 | `type()`、`len()`、断言、范围检查 | 错误是否在靠近源头的位置暴露 |
| 输出代码 | `print()`、表格或图形 | 输出是否带标签、单位和口径 |

### 推荐的最小结构

1. **输入**：只定义本单元格需要的参数，不把多个无关步骤混在一起。
2. **计算**：每个单元格完成一个动作，例如“换算小时”或“计算平均分”。
3. **检查**：先检查类型、长度和边界，再相信结果。
4. **表达**：用带标签的输出说明结果代表什么，避免只显示一个裸数字。

如果需要重跑，先重启内核，再按这个顺序从上到下执行。这样可以发现“依赖了旧变量”的隐性问题。


## 示例 1：第一个计算结果

最后一个表达式自动显示；多值或带标签的结果使用 `print()`。


```python
course_name = "Python Data Studio"
chapter_count = 75
print(course_name)
print("课程章节数:", chapter_count)
chapter_count / 3
```

## 示例 2：观察执行状态

本单元格依赖前一个单元格创建的 `chapter_count`。


```python
completed = 8
progress = completed / chapter_count
print(f"已完成 {completed} 章，进度 {progress:.1%}")
```

## 示例 3：组织一次微型分析

按“输入数据 → 指标计算 → 结果输出”组织代码。


```python
scores = [82, 91, 76, 88, 95]
average = sum(scores) / len(scores)
best = max(scores)
print(f"平均分: {average:.1f}")
print(f"最高分: {best}")
```

## 示例 4：检查输入和结果

先检查类型与长度，再输出指标，避免空列表或类型错误被忽略。


```python
scores = [82, 91, 76, 88, 95]
assert isinstance(scores, list)
assert len(scores) > 0
assert all(isinstance(score, (int, float)) for score in scores)
average = sum(scores) / len(scores)
print("样本数:", len(scores))
print("平均分:", round(average, 1))
```

## 示例 5：验证执行顺序

修改变量后重新运行依赖单元格，比较前后结果，理解内核中的状态变化。


```python
base_hours = 2
students = 5
total_hours = base_hours * students
print("第一次计算:", total_hours)

base_hours = 3
print("只修改变量后:", total_hours)
print("重新计算后:", base_hours * students)
```

## 示例 6：把参数和报告分开

把业务参数放在一个字典里，计算和输出只读取参数，修改输入时不用到处找数字。


```python
plan = {
    "course": "Python数据分析",
    "weeks": 4,
    "hours_per_week": 3,
}
total_hours = plan["weeks"] * plan["hours_per_week"]
report = (
    f"{plan['course']}：{plan['weeks']} 周 × "
    f"每周 {plan['hours_per_week']} 小时 = {total_hours} 小时"
)
print(report)
```

## 示例 7：检查类型、单位和边界

数值正确不等于口径正确；同时检查类型、单位和允许范围。


```python
days = 14
minutes_per_day = 45
hours_per_day = minutes_per_day / 60

assert isinstance(days, int)
assert isinstance(minutes_per_day, int)
assert days > 0 and 0 < minutes_per_day <= 24 * 60
print("天数类型:", type(days).__name__)
print(f"每天学习: {hours_per_day:.2f} 小时")
```

## 示例 8：受控地观察常见报错

先让错误以可读方式出现，再补上缺失变量；不要用 `try/except` 静默吞掉所有异常。


```python
try:
    print(total_minutes)
except NameError as error:
    print("变量尚未定义:", type(error).__name__)

days = 14
minutes_per_day = 45
total_minutes = days * minutes_per_day
print("补齐变量后，总分钟数:", total_minutes)
```

## 示例 9：记录一次可复核输出

输出中同时保留输入、计算口径和结果，别人才能判断数字是否算对。


```python
inputs = {"订单数": 128, "完成订单数": 96}
completion_rate = inputs["完成订单数"] / inputs["订单数"]
print("输入:", inputs)
print(f"完成率 = 完成订单数 / 订单数 = {completion_rate:.1%}")
assert 0 <= completion_rate <= 1
```

## 常见误区

- 依赖变量尚未创建就运行后续单元格
- 修改变量后未重新运行依赖它的单元格
- 只保留结果，不保留输入和计算过程
- 把分钟、小时、金额等不同单位混在同一个变量里
- 用一个很大的 try/except 把真正的错误隐藏起来
