### 1.1.1 什么是程序

程序是“数据 + 指令”的有序集合：数据是程序处理的对象（数字、文本、表格），指令是告诉计算机“做什么、怎么做”的步骤。（可以把它想成一份菜谱：数据是食材，指令是“先切、再炒、后装盘”的步骤，照做就有结果。）程序执行的过程，就是计算机按顺序执行这些指令的过程。

例如“计算本月总支出”可以拆解为三条指令：

```text
1. 读取所有支出记录（数据）
2. 对金额求和（处理）
3. 输出结果（结果）
```

### 1.1.2 Python 与解释器

Python 是一种高级编程语言。高级语言编写的源程序不能直接被计算机执行，需要由**解释器**逐行翻译成机器指令。常见的两种文件形式：

| 文件类型 | 说明 | 使用场景 |
| --- | --- | --- |
| .py | 纯代码源文件，整体运行 | 程序开发、部署 |
| .ipynb | Notebook 文档，文字与代码分单元格 | 教学、实验、数据分析报告 |

本课程使用 .ipynb：每个**单元格（Cell）**可以单独运行，适合边讲解边验证。

### 1.1.3 运行环境：内核与输出

Notebook 中真正执行 Python 代码的是一个后台进程，称为**内核（Kernel）**：

- 内核保存运行过程中产生的变量；
- 单元格按“从上到下”的顺序执行，后运行的单元格可以访问先运行单元格定义的变量；
- 重启内核会清空所有变量，必须重新按顺序运行。

> **易错点**：只运行中间的单元格而跳过前面的准备步骤，是初学者最常见的错误来源。

### 1.4.1 print()：带格式的输出

**作用**：把结果输出到屏幕。**语法**：print(值1, 值2, ..., sep=分隔符, end=结尾)。

- sep：多个值之间的分隔符，默认空格；
- end：输出结尾，默认换行。



````python
date = "2026-08-06"
category = "餐饮"
amount = 35.5
print(date, category, amount, sep=" | ")
print("金额：", end="")
print(amount)
````

**解读**：第一行用竖线分隔三个字段，适合表格化输出；第二、三行通过 end 为空字符串把两次输出连在同一行。


### 1.4.2 type() 与 isinstance()：确认数据类型

**作用**：type(值) 返回值的实际类型；isinstance(值, 类型) 返回布尔值，判断值是否属于某类型。

> **易错点**：在 Python 中布尔值 True 是整数 1 的子类，因此 isinstance(True, int) 为 True。金额校验时不能只判断整数类型。



````python
amount = 35.5
raw_amount = "35.5"
print(type(amount))  # <class float>
print(type(raw_amount))  # <class str>
print(isinstance(amount, (int, float)))  # True
print(isinstance(raw_amount, (int, float)))  # False
print(isinstance(True, int))  # True，布尔是整数的子类
````

### 1.4.3 dir() / help() / callable()：现场查阅

**作用**：dir(对象) 列出可用名字；callable(名字) 判断能否像函数一样调用；help(函数) 查看详细说明。

课程不会穷举所有方法，**学会查比死记更重要**。



````python
public_names = [name for name in dir(str) if not name.startswith("_")]
print("字符串公开名称数量：", len(public_names))
print("前 8 个：", public_names[:8])
print("strip 能调用吗：", callable(str.strip))
print("upper 能调用吗：", callable(str.upper))
````

## 1.5 易错点提醒

第一次上机最容易犯的三个错误，先看再写：

## 教学实验：边界条件

这一组实验专门训练“观察一个结果 → 只改一个变量 → 解释变化”。先运行第一个代码单元格，再运行第二个。


````python
orders = [
    {"order_id": "A01", "amount": 280},
    {"order_id": "A02", "amount": 300},
    {"order_id": "A03", "amount": 520},
]
threshold = 300
for order in orders:
    label = "达到门槛" if order["amount"] >= threshold else "未达到门槛"
    print(order["order_id"], order["amount"], label)
````

### 第一个结果怎么读

这里的重点不是记住 `if`，而是观察 `>=` 如何处理刚好等于 300 的记录。边界条件必须和业务规则保持一致。

## 错误恢复：类型转换失败怎么办

真实数据和真实代码都会出问题。本节先观察问题，再用一个明确的检查或修复步骤恢复运行。


````python
amount_text = "128.5"
try:
    amount = int(amount_text)
except ValueError as error:
    print("第一次转换失败：", type(error).__name__)
    amount = float(amount_text)
print("可以继续使用的金额：", amount)
````

### 错误恢复步骤

1. 先看错误类型、字段或数据形状。
2. 判断问题发生在输入、处理中间结果还是输出。
3. 修复后重新检查结果，而不是只让代码不报错。

先读错误类型，再决定修复方法。这里不是盲目忽略错误，而是明确知道整数转换不适合带小数的文本。
