## 32.3 第一个上手示例



先感受画一张最基础的折线图。运行下方代码，看看"画布 + 一条线 + 标签"的最小结构。


```python
import matplotlib.pyplot as plt

import numpy as np


months = ["1月", "2月", "3月", "4月", "5月"]

sales = [120, 150, 180, 210, 160]


fig, ax = plt.subplots()  # 创建画布+坐标系

ax.plot(months, sales, marker="o")  # 画折线

ax.set(title="月度销量", xlabel="月份", ylabel="销量（件）")  # 加标签

plt.show()
```

## 教学实验：图表只改一个编码

这一组实验专门训练“观察一个结果 → 只改一个变量 → 解释变化”。先运行第一个代码单元格，再运行第二个。


```python
import matplotlib.pyplot as plt

months = ["1月", "2月", "3月", "4月"]
sales = [120, 150, 138, 190]
fig, ax = plt.subplots(figsize=(7, 3.5))
ax.plot(months, sales, marker="o")
ax.set_title("月度销售额")
ax.set_ylabel("销售额（万元）")
ax.grid(alpha=0.25)
plt.show()
```

### 第一个结果怎么读

标题、坐标轴和单位让读者知道图表回答什么问题。没有这些文字，图形即使画出来也不完整。

请记录：输入是什么、输出是什么、输出支持了哪一个结论。


```python
fig, ax = plt.subplots(figsize=(7, 3.5))
ax.bar(months, sales, color="#2563EB")
ax.axhline(
    sum(sales) / len(sales), color="#DC2626", linestyle="--", label="平均值"
)
ax.set_title("月度销售额与平均值")
ax.set_ylabel("销售额（万元）")
ax.legend()
plt.show()
```

### 第二个结果怎么读

## 错误恢复：图表能画出但读不懂怎么办

真实数据和真实代码都会出问题。本节先观察问题，再用一个明确的检查或修复步骤恢复运行。


```python
import matplotlib.pyplot as plt

months = ["1月", "2月", "3月"]
sales = [120, 150, 138]
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(months, sales, marker="o")
ax.set_title("月度销售额")
ax.set_xlabel("月份")
ax.set_ylabel("销售额（万元）")
ax.grid(alpha=0.25)
plt.show()
```
