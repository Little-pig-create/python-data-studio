## 32.3 第一个上手示例



先感受画一张最基础的折线图。运行下方代码，看看"画布 + 一条线 + 标签"的最小结构。


```python
import matplotlib.pyplot as plt

import numpy as np


months = ["1月", "2月", "3月", "4月", "5月"]

sales = [120, 150, 180, 210, 160]


fig, ax = plt.subplots()  # 创建画布+坐标系

ax.plot(months, sales, marker="o")  # 画折线

ax.set(title="月度销量", xlabel="月份", ylabel="销量（件）")  # 加标签

plt.show()
```

```python
import matplotlib.pyplot as plt

days = ["周一", "周二", "周三", "周四", "周五"]
sales = [120, 138, 132, 155, 168]

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(days, sales, marker="o", linewidth=2)
ax.set_title("本周销售趋势")
ax.set_xlabel("日期")
ax.set_ylabel("销售额")
ax.grid(alpha=0.25)

plt.show()
```
