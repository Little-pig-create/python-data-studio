## 5.6 易错点提醒



### 5.6.1 count、index 与成员检查

与列表一致：count 统计次数、index 返回第一次出现位置、in 检查成员。先 in 后 index 可避免 ValueError。



```python
schema = ("date", "type", "category", "amount", "note")
print("amount" in schema)
print(schema.count("amount"))
print(schema.index("category"))
if "merchant" not in schema:
    print("当前结构没有 merchant 字段")
```

**输出解读**：成员检查 True、出现 1 次、索引 2；不存在的字段用 in 提前判断。


### 5.6.2 打包、解包与星号接收

多个值赋值给一个变量是**打包**（形成元组）；元组赋给多个变量是**解包**。星号变量接收剩余元素（列表），可用于固定首尾字段、中间不定长的场景。



```python
record = "2026-08-06", "支出", "餐饮", 35.5, "午餐"
date, record_type, *details = record
*main_fields, note = record
print(date, record_type)
print(details)
print(main_fields)
print(note)
```

**输出解读**：省略括号打包成元组；details 收集从第三个字段起的全部元素；main_fields 收集除最后一个外的全部字段。


### 5.6.3 用元组做字典键

字典键必须**可哈希**（不可变）。元组不可变，因此可以作字典键——适合表达“（月份, 分类）→ 总额”这类复合键；列表可变，不能作键。
                    
> 前置提示：此处先演示 (月份, 分类) 元组作为字典键；字典的系统知识在第 6 章学习，现在只需观察“元组可以作为键”。



```python
monthly_totals = {
    ("2026-08", "餐饮"): 320.5,
    ("2026-08", "交通"): 88.0,
    ("2026-09", "餐饮"): 210.0,
}
key = ("2026-08", "餐饮")
print(monthly_totals[key])
for (month, category), total in monthly_totals.items():
    print(month, category, total)
```

**输出解读**：复合键（月份, 分类）直接查总额；遍历 items 时先解包键元组再解包值。


### 5.6.4 namedtuple：保留元组，也给字段起名字

namedtuple 创建的类既有元组的轻量与不可变，又支持按字段名访问（record.amount）与按索引访问（record[3]），还能像元组一样解包。



```python
from collections import namedtuple

Record = namedtuple(
    "Record",
    ["date", "record_type", "category", "amount", "note"],
)
record = Record("2026-08-06", "支出", "餐饮", 35.5, "午餐")
print(record.amount)
print(record[3])
date, record_type, category, amount, note = record
print(date, category, amount)
```

**输出解读**：record.amount 按名访问、record[3] 按索引访问、解包兼容——namedtuple 兼得元组与字段名的优点。


### 5.6.5 可变与不可变：为什么改 a 会动 b

Python 中可变对象（列表、字典）被多个名字引用时，它们指向**同一个**对象——改一个名字，另一个名字也会看到变化；不可变对象（数字、字符串、元组）不存在这个问题。

- a = b = []：a 与 b 指向同一个空列表，a.append(1) 后 b 也看到 [1]；
- 元组不可变，因此 t2 = t1 后再给 t1 赋新值只是换指向，不会影响 t2；
- copy() 只复制外层容器，内层可变对象仍共享（浅复制）。

> **为什么重要**：它解释了“把列表传给函数，函数内部的修改会影响外部”这一初学者最常困惑的现象。



```python
# 可变对象的共享引用
shared = []
a = shared
b = shared
a.append(35.5)
print("b 也看到了：", b)  # 同一个可变对象，a 改了 b 也变

# 不可变对象没有“改动另见”的问题
t1 = ("餐饮", 35.5)
t2 = t1
print("t1 == t2：", t1 == t2)

# 浅复制：外层独立，内层仍共享
nested = [[1]]
shallow = nested.copy()
shallow[0].append(2)
print("内层共享：", nested, shallow)
```
