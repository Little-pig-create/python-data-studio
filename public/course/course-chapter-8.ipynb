## 示例 1：定义与返回

函数把计算与具体输入分离。


```python
def calculate_total(price, quantity, discount=1.0):
    """计算折扣后的订单总额。"""
    return price * quantity * discount

amount = calculate_total(299, 3, discount=0.9)
print(f"订单金额: {amount:.2f}")
```

## 示例 2：多返回值与作用域

Python实际返回一个元组，可以直接解包。


```python
def summarize(values):
    total = sum(values)
    average = total / len(values)
    return total, average, min(values), max(values)

sales = [120, 150, 180, 210]
total, average, minimum, maximum = summarize(sales)
print(total, average, minimum, maximum)
```

## 示例 3：Lambda与排序键

Lambda适合短小的一次性函数，复杂逻辑仍应使用def。


```python
orders = [
    {"id": "A1", "amount": 320},
    {"id": "A2", "amount": 880},
    {"id": "A3", "amount": 460},
]
ranked = sorted(orders, key=lambda item: item["amount"], reverse=True)
print(ranked)
```

## 常见误区

- 函数内部只print却期望外部获得返回值
- 修改全局变量造成隐藏状态
- 用Lambda承载多步复杂逻辑
