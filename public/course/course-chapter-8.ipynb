## 8.6 易错点提醒




### 8.6.1 真值、any 与 all

Python 中空字符串、0、空列表、None 为假值，其余为真值。any(列表) 任一为真即 True；all(列表) 全部为真才 True。




```python
record = {
    "date": "2026-08-06",
    "type": "支出",
    "category": "餐饮",
    "amount": 35.5,
    "note": "",
}
required = [
    record["date"],
    record["type"],
    record["category"],
    record["amount"] > 0,
]
print("必填项都有效：", all(required))
print("存在空值：", any(not value for value in record.values()))
```

**输出解读**：必填项全部有效；note 为空字符串（假值），any 检测到空值返回 True。


### 8.6.2 and 与 or 会短路（还会返回对象值）

**讲解**：and 左侧为假时不再计算右侧；or 左侧为真时不再计算右侧。除了"短路"，更重要的是它们**返回的是其中一个对象**，不是 True/False：

- `or`：左侧为真就返回左侧，否则返回右侧——经典的"取默认值"写法 `name = user.get("name") or "未知"`；
- `and`：左侧为假就叫返回左侧，否则返回右侧——常用于"非空才继续" `note and note.strip()`；
- 因为假值包括 `0`、`""`、`[]`、`None`，所以 `0 or 100` 会得到 100。

利用短路可以安全地组合检查：`note is not None and note.strip() != ""` 不会在 None 上调用 strip。

> **口诀**：or 取"真保留"，and 取"假早退"；返回的是对象，不是简单的对错。



```python
# 短路返回对象值：常用默认值写法
user = {"name": "小明"}
name = user.get("name") or "未知用户"  # 缺省时给默认
score = "" or 0 or 100  # 取第一个'真值'
note = None
clean = note and note.strip()  # None 时安全返回 None
print("name:", name)
print("score:", score)
print("clean:", clean)
```

```python
note = None
has_note = note is not None and note.strip() != ""
print(has_note)

category = ""
display_category = category or "其他"
print(display_category)

amount_text = "35.5"
can_convert = amount_text and amount_text.replace(".", "", 1).isdigit()
print(can_convert)
```

**输出解读**：短路保证 None 安全；or 为假值提供默认值；字符串为空时不再执行 isdigit 检查。


### 8.6.3 早返回比层层嵌套更清楚

校验函数中，每条规则失败立即 return，避免 if/else 层层嵌套。返回 (是否通过, 原因) 元组便于调用方直接使用。




```python
def check_record(record):
    if not record.get("date"):
        return False, "缺少日期"
    if record.get("type") not in {"收入", "支出"}:
        return False, "类型不合法"
    if record.get("amount", 0) <= 0:
        return False, "金额必须大于 0"
    return True, "记录有效"


samples = [
    {"type": "支出", "amount": 35.5},
    {"date": "2026-08-06", "type": "未知", "amount": 35.5},
]
for sample in samples:
    print(check_record(sample))
```

**输出解读**：每条规则失败立即返回明确原因；合法记录返回 (True, "记录有效")。


### 8.6.4 字典分派：固定菜单的另一种写法

选项与动作一一对应时，用字典 + get 比一长串 if/elif 更清晰：actions.get(choice, "未知选项")。




```python
actions = {
    "1": "添加账目",
    "2": "查看账目",
    "3": "查询账目",
    "4": "删除账目",
    "5": "统计汇总",
    "6": "退出",
}
for choice in ["3", "9"]:
    action = actions.get(choice, "未知选项")
    print(choice, action)
```

**输出解读**：选项 3 命中“查询账目”；选项 9 不在字典中，get 返回默认值“未知选项”。


### 8.6.5 用边界表检查 > 和 >=

边界是最容易出错的地方。把边界用例组织成表（值, 期望标签）逐个验证，比拍脑袋测试更可靠。




```python
def amount_level(amount):
    if amount <= 0:
        return "拒绝"
    if amount >= 5000:
        return "需复核"
    return "正常"


cases = [
    (-0.01, "拒绝"),
    (0, "拒绝"),
    (0.01, "正常"),
    (4999.99, "正常"),
    (5000, "需复核"),
]
for value, expected in cases:
    actual = amount_level(value)
    print(value, actual, actual == expected)
```

**输出解读**：五个边界用例全部通过——0 归拒绝、0.01 归正常、4999.99 归正常、5000 归需复核。边界表让规则可验证。
