## 本章场景



真实的数据大多以"表格"形式存在：CSV、Excel、数据库导出，每一行是一条记录、每一列是一个字段。手工在 Excel 里点筛选、拖汇总既慢又难复用。



**Pandas** 是 Python 处理表格数据的核心库。它把一张表装进 **DataFrame**（二维表格）和 **Series**（一列数据），让你用几行代码完成筛选、排序、清洗、聚合——这些操作在 Excel 里要反复点击。它是整个数据分析流里"备料"的一步：先把数据变成干净、可计算的表格，后面才能画图、建模。


## 本章目标

学完本章，你将能够：

- **理解**：理解 Pandas 的定位：用 Series/DataFrame 组织表格数据，并串联起后续所有数据步骤。
- **操作**：能创建 DataFrame、读入数据并做基础的取列、筛选与汇总。
- **迁移**：能把一批订单记录整理成表格并算出地区的汇总经营结果。


## 22.1 这个模块解决什么问题



**背景引入**：上一阶段用 NumPy 数组处理数值，但真实数据还带着列名、且类型混杂（文字、日期、数字）。你需要一个能看懂"表格结构"的容器。



- **看懂结构**：DataFrame 有行索引、列名，一眼看清多少行多少列、每列什么类型；

- **按条件取数**：用 loc / iloc 按标签或位置选取，用布尔条件筛选；

- **清洗与汇总**：处理缺失、重复、异常，再分组聚合出结论；

- **读写文件**：一键读取 CSV / Excel，处理完再导出去。

## 22.3 第一个上手示例



先感受一下"看懂一张表"。运行下方代码：读入几行销售记录，看形状/类型，再按条件筛出金额超过 200 的行。


```python
import pandas as pd


orders = pd.DataFrame(
    {
        "订单": ["A001", "A002", "A003", "A004"],
        "分类": ["餐饮", "购物", "餐饮", "交通"],
        "金额": [35.5, 299.0, 128.0, 18.0],
    }
)

print("形状:", orders.shape)

print("列类型:\n", orders.dtypes)

print("\n金额超过 200 的行:")

print(orders[orders["金额"] > 200])
```

```python
import pandas as pd

raw_orders = {
    "order_id": ["A01", "A02", "A03", "A04"],
    "region": ["华东", "华南", "华东", "华北"],
    "amount": [120, 80, 150, 90],
}

orders = pd.DataFrame(raw_orders)
regional_sales = (
    orders.groupby("region")["amount"]
    .sum()
    .sort_values(ascending=False)
)

print(regional_sales)
```
