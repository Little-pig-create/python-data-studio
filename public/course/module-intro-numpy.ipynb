## 16.1 这个模块解决什么问题



**背景引入**：列表能装数据，但不能高效"算"。你需要一个能整批计算、能看形状、能取一部分、能做统计的数值容器。



- **整批计算**：不用循环，直接 `数组 + 数组`、`数组 * 常量`；

- **形状感知**：知道数据是几行几列（shape）、几维（ndim）、什么类型（dtype）；

- **高效取数**：按行、按列、按条件筛选，秒级完成。

## 16.3 第一个上手示例



先感受一下数组的"整批计算"。运行下方代码：把每天的销量统一上浮 10%，并算出总和与平均值。


```python
import numpy as np


sales = np.array([120, 150, 180, 210, 160])  # 一周五天的销量

boosted = sales * 1.10  # 整批上浮 10%

print("原销量:", sales)

print("上浮后:", boosted.round(1))

print("合计:", boosted.sum(), "平均:", boosted.mean())
```

## 教学实验：axis与布尔筛选

这一组实验专门训练“观察一个结果 → 只改一个变量 → 解释变化”。先运行第一个代码单元格，再运行第二个。


```python
import numpy as np

matrix = np.arange(1, 13).reshape(3, 4)
print("原数组：\n", matrix)
print("每行合计：", matrix.sum(axis=1))
print("每列合计：", matrix.sum(axis=0))
```

### 第一个结果怎么读

`axis=1` 保留行，沿列方向计算；`axis=0` 保留列，沿行方向计算。先看 shape，再解释结果长度。

请记录：输入是什么、输出是什么、输出支持了哪一个结论。


```python
even = matrix[matrix % 2 == 0]
print("偶数：", even)
print("偶数数量：", even.size)
print("偶数平均值：", even.mean())
```

### 第二个结果怎么读

第二个实验不改原数组，而是用布尔条件筛选新数组。请思考：如果条件改成 `matrix > 8`，输出会怎样变化？

## 错误恢复：数组形状不匹配怎么办

真实数据和真实代码都会出问题。本节先观察问题，再用一个明确的检查或修复步骤恢复运行。


```python
import numpy as np

matrix = np.arange(6).reshape(2, 3)
try:
    result = matrix + np.array([10, 20])
except ValueError as error:
    print("形状问题：", type(error).__name__)
    result = matrix + np.array([10, 20, 30])
print("修复后的结果：")
print(result)
```

### 错误恢复步骤

1. 先看错误类型、字段或数据形状。
2. 判断问题发生在输入、处理中间结果还是输出。
3. 修复后重新检查结果，而不是只让代码不报错。
