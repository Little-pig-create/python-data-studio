## 本章场景



上一阶段你用 Python 列表存了一笔笔账目。现在假设要处理一整年的销售数据——几十万条单价 × 数量：用列表一个个乘、一个个算，写一堆 for 循环，跑得又慢又容易出错。



**NumPy** 就是为"整批数值计算"而生的：它把一列数存成一个 **ndarray 数组**，一次加减乘除对整批数据同时生效，速度比 Python 列表快一个数量级。它是数据分析领域一切数值运算的地基，也是后边 Pandas 表格的底层引擎。


## 本章目标

学完本章，你将能够：

- **理解**：理解 NumPy 在数据分析中的定位（多维数组与向量化计算），以及它和 Python 列表的区别。
- **操作**：能创建 ndarray、做索引/切片/形状操作与基础运算。
- **迁移**：能用 NumPy 把一批经营数据转成数组并做批量计算。


## 16.1 这个模块解决什么问题



**背景引入**：列表能装数据，但不能高效"算"。你需要一个能整批计算、能看形状、能取一部分、能做统计的数值容器。



- **整批计算**：不用循环，直接 `数组 + 数组`、`数组 * 常量`；

- **形状感知**：知道数据是几行几列（shape）、几维（ndim）、什么类型（dtype）；

- **高效取数**：按行、按列、按条件筛选，秒级完成。


## 16.2 学习地图



这个模块从零到一看懂数组，共 4 个技术章节：



1. **16.1 数组基础**：怎么创建数组、看懂 shape / ndim / dtype；

2. **16.2 索引、切片与筛选**：高效取出某一行、某一段、满足条件的值；

3. **16.3 形状、合并与拆分**：给数据换形状、拼多份、切几块；

4. **16.4 统计计算与随机抽样**：按轴算汇总、分位数、异常阈值、随机模拟。



它前面是 Python 基础，后面是 **Pandas**——DataFrame 底层的每一列就是一个 NumPy 数组。先把数组这张牌打好，表格分析才会顺手。


## 16.3 第一个上手示例



先感受一下数组的"整批计算"。运行下方代码：把每天的销量统一上浮 10%，并算出总和与平均值。


```python
import numpy as np


sales = np.array([120, 150, 180, 210, 160])  # 一周五天的销量

boosted = sales * 1.10  # 整批上浮 10%

print("原销量:", sales)

print("上浮后:", boosted.round(1))

print("合计:", boosted.sum(), "平均:", boosted.mean())
```
