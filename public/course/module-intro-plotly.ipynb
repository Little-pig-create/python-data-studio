## 本章场景



前面画的是静态图——输出就是一张定格图片。当你想在页面上"鼠标悬停看清数值、拖拽缩放、点击隐藏某条线"时，静态图就办不到了。



**Plotly** 画的是**交互式图表**：悬停能看数据、缩放、联动，非常适合运营看板、数据报告、演示汇报。它用 `px.` 一行生成图（快速出图），需要精细控制时再用 `go.`（子图、3D、地图）。学会它，你就能做出"能交互的数据看板"。

- **理解**：理解 Plotly 面向交互式可视化的定位与图对象（Figure）概念。
- **操作**：能用 plotly.express 画一个基础交互图并查看 hover。
- **迁移**：能把一份经营表变成可交互探索的图表用于汇报。


## 64.1 这个模块解决什么问题



**背景引入**：看静态报表时，想看清某根折线背后的具体数值，还得回表格里翻；而交互图悬停即现。



- **交互默认内置**：悬停看数、图例点击显隐、拖拽缩放；

- **px 快速出图**：`px.line`、`px.bar` 一行生成；

- **go 精细控制**：子图组合、坐标轴联动、地图、3D。

## 64.3 第一个上手示例



感受交互式折线图。运行下方代码：先出一张线图，再试着把鼠标悬停在点上看看数值。


```python
import plotly.express as px

months = ["1月", "2月", "3月", "4月", "5月"]
sales = [120, 150, 180, 210, 160]

fig = px.line(x=months, y=sales, title="月度销量（交互）")
fig.show()
```

```python
import pandas as pd
import plotly.express as px

weekly = pd.DataFrame({
    "date": pd.date_range("2026-08-03", periods=5, freq="D"),
    "sales": [120, 138, 132, 155, 168],
    "target": [130, 130, 140, 150, 160],
})

fig = px.line(
    weekly,
    x="date",
    y="sales",
    markers=True,
    hover_data=["target"],
    title="本周销售与目标核对",
)
fig.show()
```
