## 核心概念

- 代码单元格负责计算，Markdown单元格负责解释。
- 执行编号表示运行顺序，不等于单元格在页面中的位置。
- 内核保存已执行变量；重启内核后需要重新运行依赖单元格。


## 示例 1：第一个计算结果

表达式最后一行会直接显示，print适合同时输出多个说明性结果。


```python
course_name = "Python Data Studio"
chapter_count = 75
print(course_name)
print("课程章节数:", chapter_count)
chapter_count / 3
```

## 示例 2：观察执行状态

后续单元格可以访问前面已经创建的变量。


```python
completed = 8
progress = completed / chapter_count
print(f"已完成 {completed} 章，进度 {progress:.1%}")
```

## 示例 3：组织一次微型分析

先准备数据，再计算指标，最后输出结论。


```python
scores = [82, 91, 76, 88, 95]
average = sum(scores) / len(scores)
best = max(scores)
print(f"平均分: {average:.1f}")
print(f"最高分: {best}")
```

## 常见误区

- 跳过数据准备单元格后直接运行后续代码
- 只看图表而不记录计算口径
- 修改变量后忘记重新运行受影响的单元格
