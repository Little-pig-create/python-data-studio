## 1. 什么是变量？

变量可以理解为一个用于引用数据的名称。使用变量后，可以在程序中重复使用或更新数据。

基本形式：
```python
变量名 = 数据
```

````python
student_name = "小明"
score = 92
passed = score >= 60

print(student_name)
print(score)
print(passed)
````

## 2. 变量的重新赋值

同一个变量可以在程序运行过程中保存不同的数据。


````python
temperature = 25
print(temperature)

temperature = 31
print(temperature)
````

## 3. 标识符命名规则

Python标识符主要包括变量名、函数名等。命名时应遵守以下规则：

1. 只能包含字母、数字和下划线。
2. 不能以数字开头。
3. 不能使用Python关键字。
4. 区分大小写。
5. 建议使用具有实际含义的英文单词或拼音。

推荐使用小写字母加下划线的命名方式，例如 `total_sales`。


````python
# 推荐命名
user_name = "Alice"
total_sales = 1280.5
order_count = 36

# Python区分大小写
value = 10
Value = 20
print(value, Value)
````

## 4. 常见错误示例

下面的命名不符合Python标识符规则：

- `2nd_place`：不能以数字开头
- `user-name`：连字符会被解释为减号
- `class`：属于Python关键字
- `user name`：变量名中不能包含空格


````python
import keyword

print(keyword.iskeyword("class"))
print(keyword.iskeyword("score"))
print(keyword.kwlist)
````

## 5. 查看变量类型

可以使用 `type()` 查看变量当前引用的数据类型。


````python
name = "Python"
count = 10
price = 19.9
is_open = True

print(type(name))
print(type(count))
print(type(price))
print(type(is_open))
````
