## 1. 什么是变量？

变量可以理解为一个用于引用数据的名称。使用变量后，可以在程序中重复使用或更新数据。

基本形式：
```python
变量名 = 数据
```


````python
student_name = "小明"
score = 92
passed = score >= 60

print(student_name)
print(score)
print(passed)
````

## 2. 变量的重新赋值

同一个变量可以在程序运行过程中保存不同的数据。


````python
temperature = 25
print(temperature)

temperature = 31
print(temperature)
````

## 3. 标识符命名规则

Python标识符主要包括变量名、函数名等。命名时应遵守以下规则：

1. 只能包含字母、数字和下划线。
2. 不能以数字开头。
3. 不能使用Python关键字。
4. 区分大小写。
5. 建议使用具有实际含义的英文单词或拼音。

推荐使用小写字母加下划线的命名方式，例如 `total_sales`。


````python
# 推荐命名
user_name = "Alice"
total_sales = 1280.5
order_count = 36

# Python区分大小写
value = 10
Value = 20
print(value, Value)
````

## 4. 常见错误示例

下面的命名不符合Python标识符规则：

- `2nd_place`：不能以数字开头
- `user-name`：连字符会被解释为减号
- `class`：属于Python关键字
- `user name`：变量名中不能包含空格


````python
import keyword

print(keyword.iskeyword("class"))
print(keyword.iskeyword("score"))
print(keyword.kwlist)
````

## 5. 查看变量类型

可以使用 `type()` 查看变量当前引用的数据类型。


````python
name = "Python"
count = 10
price = 19.9
is_open = True

print(type(name))
print(type(count))
print(type(price))
print(type(is_open))
````

## 教学实验：边界条件

这一组实验专门训练“观察一个结果 → 只改一个变量 → 解释变化”。先运行第一个代码单元格，再运行第二个。


````python
orders = [
    {"order_id": "A01", "amount": 280},
    {"order_id": "A02", "amount": 300},
    {"order_id": "A03", "amount": 520},
]
threshold = 300
for order in orders:
    label = "达到门槛" if order["amount"] >= threshold else "未达到门槛"
    print(order["order_id"], order["amount"], label)
````

### 第一个结果怎么读

这里的重点不是记住 `if`，而是观察 `>=` 如何处理刚好等于 300 的记录。边界条件必须和业务规则保持一致。

请记录：输入是什么、输出是什么、输出支持了哪一个结论。


````python
threshold = 500
for order in orders:
    label = "重点关注" if order["amount"] >= threshold else "普通订单"
    print(order["order_id"], "->", label)
print("重点订单数：", sum(order["amount"] >= threshold for order in orders))
````

### 第二个结果怎么读

只把门槛从 300 改成 500，再比较标签和数量变化。这个实验训练的是“改一个输入，解释一个输出”。

## 错误恢复：类型转换失败怎么办

真实数据和真实代码都会出问题。本节先观察问题，再用一个明确的检查或修复步骤恢复运行。


````python
amount_text = "128.5"
try:
    amount = int(amount_text)
except ValueError as error:
    print("第一次转换失败：", type(error).__name__)
    amount = float(amount_text)
print("可以继续使用的金额：", amount)
````

### 错误恢复步骤

1. 先看错误类型、字段或数据形状。
2. 判断问题发生在输入、处理中间结果还是输出。
3. 修复后重新检查结果，而不是只让代码不报错。

先读错误类型，再决定修复方法。这里不是盲目忽略错误，而是明确知道整数转换不适合带小数的文本。
