# 2.1 变量与标识符

变量可以理解为一个用于引用数据的名称。基本形式：

```python
变量名 = 数据
```

标识符命名规则：

- 只能包含字母、数字和下划线；
- 不能以数字开头；
- 不能使用Python关键字；
- 区分大小写；
- 不能包含空格和连字符；
- 推荐使用有意义的蛇形命名。


````python
student_name = "小明"
student_age = 20
student_score = 92.5
is_passed = True

print(student_name)
print(student_age)
print(student_score)
print(is_passed)
````

## 变量重新赋值

变量的值可以在程序运行过程中发生变化。


````python
temperature = 25
print(f"当前温度：{temperature}℃")

temperature = 31
print(f"更新后的温度：{temperature}℃")
````

## 多变量赋值


````python
x, y, z = 10, 20, 30
print(x, y, z)

a = b = c = 0
print(a, b, c)
````

## 检查关键字

Python关键字不能直接作为变量名使用。


````python
import keyword

print(keyword.iskeyword("class"))
print(keyword.iskeyword("score"))
print(keyword.kwlist)
````

# 2.2 基本数据类型

本章重点认识以下类型：

| 类型 | 示例 | 说明 |
|---|---|---|
| `int` | `20` | 整数 |
| `float` | `19.9` | 浮点数 |
| `bool` | `True` | 布尔值 |
| `str` | `"Python"` | 字符串 |
| `NoneType` | `None` | 空值 |


````python
age = 20
price = 19.9
is_open = True
name = "Python"
result = None

print(age, price, is_open, name, result)
````

## 字符串与浮点数示例


````python
first_name = "张"
last_name = "三"
full_name = first_name + last_name
print(full_name)

result = 0.1 + 0.2
print(result)
print(f"格式化结果：{result:.2f}")
````

# 2.3 类型查看与类型转换


````python
values = ["Python", 10, 19.9, True, None]

for value in values:
    print(value, type(value))
````

## 常用类型转换


````python
age_text = "20"
age = int(age_text)

price_text = "19.9"
price = float(price_text)

number = 100
number_text = str(number)

print(age, type(age))
print(price, type(price))
print(number_text, type(number_text))
````

## 布尔类型转换


````python
print(bool(0))
print(bool(1))
print(bool(""))
print(bool("Python"))
print(bool(None))
````

# 2.4 运算符

## 算术运算符

`+`、`-`、`*`、`/`、`//`、`%`、`**`。


````python
a = 5
b = 2

print("加法：", a + b)
print("减法：", a - b)
print("乘法：", a * b)
print("除法：", a / b)
print("整除：", a // b)
print("取余：", a % b)
print("幂运算：", a ** b)
````

## 比较运算符

比较运算的结果是布尔值。


````python
score = 86

print(score == 86)
print(score != 60)
print(score > 60)
print(score < 100)
print(score >= 60)
print(score <= 100)
````

## 逻辑运算符


````python
age = 22
score = 88

print("成年且及格：", age >= 18 and score >= 60)
print("未成年或及格：", age < 18 or score >= 60)
print("未成年：", not(age >= 18))
````

## 赋值运算符与成员运算符


````python
count = 10
count += 2
count *= 3
print("count：", count)

text = "Python数据分析"
print("Python" in text)
print("Java" not in text)
````

## 教学实验：边界条件

这一组实验专门训练“观察一个结果 → 只改一个变量 → 解释变化”。先运行第一个代码单元格，再运行第二个。


````python
orders = [
    {"order_id": "A01", "amount": 280},
    {"order_id": "A02", "amount": 300},
    {"order_id": "A03", "amount": 520},
]
threshold = 300
for order in orders:
    label = "达到门槛" if order["amount"] >= threshold else "未达到门槛"
    print(order["order_id"], order["amount"], label)
````

### 第一个结果怎么读

这里的重点不是记住 `if`，而是观察 `>=` 如何处理刚好等于 300 的记录。边界条件必须和业务规则保持一致。

请记录：输入是什么、输出是什么、输出支持了哪一个结论。


````python
threshold = 500
for order in orders:
    label = "重点关注" if order["amount"] >= threshold else "普通订单"
    print(order["order_id"], "->", label)
print("重点订单数：", sum(order["amount"] >= threshold for order in orders))
````

### 第二个结果怎么读

只把门槛从 300 改成 500，再比较标签和数量变化。这个实验训练的是“改一个输入，解释一个输出”。

## 错误恢复：类型转换失败怎么办

真实数据和真实代码都会出问题。本节先观察问题，再用一个明确的检查或修复步骤恢复运行。


````python
amount_text = "128.5"
try:
    amount = int(amount_text)
except ValueError as error:
    print("第一次转换失败：", type(error).__name__)
    amount = float(amount_text)
print("可以继续使用的金额：", amount)
````

### 错误恢复步骤

1. 先看错误类型、字段或数据形状。
2. 判断问题发生在输入、处理中间结果还是输出。
3. 修复后重新检查结果，而不是只让代码不报错。

先读错误类型，再决定修复方法。这里不是盲目忽略错误，而是明确知道整数转换不适合带小数的文本。
