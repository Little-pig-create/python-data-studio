# 2.1 变量与标识符

变量可以理解为一个用于引用数据的名称。基本形式：

```python
变量名 = 数据
```

标识符命名规则：

- 只能包含字母、数字和下划线；
- 不能以数字开头；
- 不能使用Python关键字；
- 区分大小写；
- 不能包含空格和连字符；
- 推荐使用有意义的蛇形命名。

````python
student_name = "小明"
student_age = 20
student_score = 92.5
is_passed = True

print(student_name)
print(student_age)
print(student_score)
print(is_passed)
````

## 变量重新赋值

变量的值可以在程序运行过程中发生变化。

````python
temperature = 25
print(f"当前温度：{temperature}℃")

temperature = 31
print(f"更新后的温度：{temperature}℃")
````

## 多变量赋值

````python
x, y, z = 10, 20, 30
print(x, y, z)

a = b = c = 0
print(a, b, c)
````

## 检查关键字

Python关键字不能直接作为变量名使用。

````python
import keyword

print(keyword.iskeyword("class"))
print(keyword.iskeyword("score"))
print(keyword.kwlist)
````

# 2.2 基本数据类型

本章重点认识以下类型：

| 类型 | 示例 | 说明 |
|---|---|---|
| `int` | `20` | 整数 |
| `float` | `19.9` | 浮点数 |
| `bool` | `True` | 布尔值 |
| `str` | `"Python"` | 字符串 |
| `NoneType` | `None` | 空值 |

````python
age = 20
price = 19.9
is_open = True
name = "Python"
result = None

print(age, price, is_open, name, result)
````

## 字符串与浮点数示例

````python
first_name = "张"
last_name = "三"
full_name = first_name + last_name
print(full_name)

result = 0.1 + 0.2
print(result)
print(f"格式化结果：{result:.2f}")
````

# 2.3 类型查看与类型转换

````python
values = ["Python", 10, 19.9, True, None]

for value in values:
    print(value, type(value))
````

## 常用类型转换

````python
age_text = "20"
age = int(age_text)

price_text = "19.9"
price = float(price_text)

number = 100
number_text = str(number)

print(age, type(age))
print(price, type(price))
print(number_text, type(number_text))
````

## 布尔类型转换

````python
print(bool(0))
print(bool(1))
print(bool(""))
print(bool("Python"))
print(bool(None))
````

# 2.4 运算符

## 算术运算符

`+`、`-`、`*`、`/`、`//`、`%`、`**`。

````python
a = 5
b = 2

print("加法：", a + b)
print("减法：", a - b)
print("乘法：", a * b)
print("除法：", a / b)
print("整除：", a // b)
print("取余：", a % b)
print("幂运算：", a ** b)
````

## 比较运算符

比较运算的结果是布尔值。

````python
score = 86

print(score == 86)
print(score != 60)
print(score > 60)
print(score < 100)
print(score >= 60)
print(score <= 100)
````

## 逻辑运算符

````python
age = 22
score = 88

print("成年且及格：", age >= 18 and score >= 60)
print("未成年或及格：", age < 18 or score >= 60)
print("未成年：", not(age >= 18))
````

## 赋值运算符与成员运算符

````python
count = 10
count += 2
count *= 3
print("count：", count)

text = "Python数据分析"
print("Python" in text)
print("Java" not in text)
````

# 2.5 输入与格式化输出

`input()`接收的数据默认是字符串。进行数学计算前，需要先转换类型。

````python
# 运行本单元时，可以输入自己的数据
name = input("请输入姓名：")
age = int(input("请输入年龄："))

print(f"你好，{name}，你明年{age + 1}岁。")
````

## f-string格式化

````python
name = "小明"
score = 92
price = 1234567.89
rate = 0.856

print(f"姓名：{name}，成绩：{score}")
print(f"价格：{price:.2f}元")
print(f"完成率：{rate:.1%}")
print(f"销售额：{price:,.2f}元")
````

# 2.7 参考答案

````python
# 练习一
name = "小明"
age = 20
city = "上海"
job = "学生"
print(f"我叫{name}，今年{age}岁，来自{city}，是一名{job}。")

# 练习二
unit_price = 29.9
quantity = 3
discount = 0.85
final_amount = unit_price * quantity * discount
print(f"折后金额：{final_amount:.2f}元")

# 练习三
chinese, math, english = 85, 92, 78
average_score = (chinese + math + english) / 3
print(f"平均分：{average_score:.2f}，是否及格：{average_score >= 60}")

# 练习四
height = 1.75
weight = 65
bmi = weight / height ** 2
print(f"BMI：{bmi:.2f}")

# 练习五
year = 2024
is_leap_year = (year % 4 == 0 and year % 100 != 0) or year % 400 == 0
print(f"{year}年是否为闰年：{is_leap_year}")
````
