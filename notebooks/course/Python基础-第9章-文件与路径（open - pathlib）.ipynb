## 核心概念

- 文本读写应显式指定UTF-8编码。
- with块结束后自动关闭文件。
- 浏览器内核中的文件位于临时文件系统，页面重新加载后可能消失。


## 示例 1：写入和读取文本

write不会自动添加换行，需要显式提供。


```python
from pathlib import Path

file_path = Path("/tmp/course_notes.txt")
with file_path.open("w", encoding="utf-8") as file:
    file.write("Python数据分析\n")
    file.write("今日完成：字符串与函数\n")

with file_path.open("r", encoding="utf-8") as file:
    content = file.read()
print(content)
```

## 示例 2：逐行处理

逐行遍历适合较大文本，strip去除行尾换行。


```python
with file_path.open("r", encoding="utf-8") as file:
    lines = [line.strip() for line in file if line.strip()]
for number, line in enumerate(lines, start=1):
    print(number, line)
```

## 示例 3：路径信息

Path提供跨平台的路径拼接和属性访问。


```python
output_dir = Path("/tmp") / "course_outputs"
output_dir.mkdir(exist_ok=True)
report_path = output_dir / "summary.txt"
report_path.write_text("有效订单: 18\n销售额: 3280", encoding="utf-8")
print(report_path.name)
print(report_path.suffix)
print(report_path.read_text(encoding="utf-8"))
```

## 常见误区

- 依赖Windows反斜杠手工拼接路径
- 未指定编码导致中文乱码
- 以为浏览器临时文件会永久保存
