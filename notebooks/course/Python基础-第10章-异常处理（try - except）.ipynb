# 第10章 异常处理（try / except）

用异常处理管理可预期失败，并把错误转化为清晰、可恢复的信息。

- 捕获具体异常

## 核心概念

- 只捕获能够处理的具体异常。
- else在未发生异常时运行，finally无论成功失败都会运行。
- 数据校验失败时主动raise比继续计算更安全。


## 示例 1：捕获转换错误

用户或外部数据转换是常见的可预期失败。


```python
samples = ["128.5", "N/A", "360"]
values = []
for sample in samples:
    try:
        values.append(float(sample))
    except ValueError:
        print("跳过无效值:", sample)
print(values)
```

## 示例 2：else与finally

else存放只在成功时执行的逻辑。


```python
text = "42"
try:
    number = int(text)
except ValueError:
    result = None
else:
    result = number * 2
finally:
    print("转换流程结束")
print("结果:", result)
```

## 示例 3：主动校验

函数应在输入违反约束时立即给出明确错误。


```python
def conversion_rate(orders, visits):
    if visits <= 0:
        raise ValueError("访问量必须大于0")
    if orders < 0:
        raise ValueError("订单量不能为负数")
    return orders / visits

print(f"转化率: {conversion_rate(36, 420):.1%}")
```

## 常见误区

- 使用except Exception隐藏所有错误
- 捕获异常后不记录任何信息
- 把正常分支逻辑全部塞进try块
