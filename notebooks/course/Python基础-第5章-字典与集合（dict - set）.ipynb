## 核心概念

- 字典通过唯一键定位值，适合表示一条结构化记录。
- get可以为缺失键提供默认值。
- 集合无序且元素唯一，适合成员判断与去重。


## 示例 1：字典访问与更新

直接索引要求键存在，get更适合不确定字段。


```python
order = {"id": "A102", "region": "华东", "amount": 1280.5}
order["channel"] = "线上"
order["amount"] += 120
print(order["id"], order.get("coupon", 0))
for key, value in order.items():
    print(key, "=>", value)
```

## 示例 2：嵌套字典与推导式

嵌套结构要逐层访问，并对缺失层级保持谨慎。


```python
sales = {
    "华东": {"一月": 120, "二月": 150},
    "华南": {"一月": 98, "二月": 132},
}
totals = {region: sum(months.values()) for region, months in sales.items()}
print(totals)
```

## 示例 3：集合关系

集合运算直接表达共同、全部和独有成员。


```python
january_users = {"u1", "u2", "u3", "u5"}
february_users = {"u2", "u3", "u4", "u6"}
print("两月都活跃:", january_users & february_users)
print("至少活跃一次:", january_users | february_users)
print("一月独有:", january_users - february_users)
```

## 常见误区

- 访问不存在的键导致KeyError
- 使用可变对象作为字典键或集合元素
- 依赖集合的显示顺序
