# 第1章 Python与Notebook入门

掌握单元格类型、内核状态、执行顺序和结果输出。

## 本章定位

运行代码前先执行依赖单元格；需要复现结果时从上到下运行全部代码。

## 学习目标

- 区分Markdown与代码单元格
- 理解执行编号与变量状态
- 使用print查看中间结果
- 按依赖顺序组织一次简单计算


## 核心概念

- Markdown 单元格保存说明；代码单元格提交 Python 代码。
- `In [n]` 是实际执行顺序，不代表页面位置。
- 变量存活于当前内核；重启内核会清空变量和输出状态。


## 示例 1：第一个计算结果

单元格最后一个表达式自动显示；需要输出多个值或带标签的结果时使用 `print()`。


```python
course_name = "Python Data Studio"
chapter_count = 75
print(course_name)
print("课程章节数:", chapter_count)
chapter_count / 3
```

## 示例 2：观察执行状态

本单元格依赖前一个单元格创建的 `chapter_count`。


```python
completed = 8
progress = completed / chapter_count
print(f"已完成 {completed} 章，进度 {progress:.1%}")
```

## 示例 3：组织一次微型分析

按“输入数据 → 指标计算 → 结果输出”组织代码，避免混用中间变量。


```python
scores = [82, 91, 76, 88, 95]
average = sum(scores) / len(scores)
best = max(scores)
print(f"平均分: {average:.1f}")
print(f"最高分: {best}")
```

## 常见误区

- 依赖变量尚未创建就运行后续单元格
- 修改变量后未重新运行依赖它的单元格
- 只保留输出，不保留输入和计算过程


## 综合练习

1. 创建课程名称、学习天数和每日时长三个变量
2. 计算总学习时长
3. 用一句完整文本输出结果

要求：变量名与题目一致，结果保留 1 位小数。


```python
# 定义变量
course = "Python数据分析"
days = 14
minutes_per_day = 45

# 计算总学习时长（小时）
total_hours = days * minutes_per_day / 60

print(f"{course}计划共 {days} 天，预计学习 {total_hours:.1f} 小时")
```

```python
course = "Python数据分析"
days = 14
minutes_per_day = 45
total_hours = days * minutes_per_day / 60
print(f"{course}计划共 {days} 天，预计学习 {total_hours:.1f} 小时")

# 自检
assert total_hours == 10.5, "检查时长计算：应该是 days * minutes_per_day / 60"
assert isinstance(total_hours, float), "total_hours 应该是浮点数"
```
