## 核心概念

- reshape前后元素总数必须一致。

- 合并前除目标轴外的其他维度必须匹配。


## 示例 1：形状变换与转置

reshape只改变视图结构，不改变元素顺序。


```python
import numpy as np

values = np.arange(1, 13)
matrix = values.reshape(3, 4)
print(matrix)
print("转置:\n", matrix.T)
print("展平:", matrix.ravel())
```

## 示例 2：数组合并

vstack按行叠加，hstack按列拼接。


```python
first = np.array([[1, 2], [3, 4]])
second = np.array([[5, 6], [7, 8]])
print("按行:\n", np.vstack([first, second]))
print("按列:\n", np.hstack([first, second]))
print("新增维度:\n", np.stack([first, second], axis=0).shape)
```

## 示例 3：数组拆分

split要求能够等分，array_split允许不等分。


```python
monthly = np.arange(1, 13)
quarters = np.split(monthly, 4)
uneven = np.array_split(monthly, 5)
print("季度:", quarters)
print("不等分长度:", [len(part) for part in uneven])
```

## 示例 4：整理数组并查看每行合计

这一组例子只处理一个小问题。先运行代码，再逐行对照拆解说明。


```python
import numpy as np

daily_sales = np.array([120, 150, 180, 210, 195, 230])
table = daily_sales.reshape(2, 3)
print("数组：\n", table)
print("形状：", table.shape)
print("每行合计：", table.sum(axis=1))
```

### 逐步拆解

reshape 只改变排列方式，不改变元素总数；axis=1 表示沿每一行计算。

建议第一次运行后只改一个输入值，再观察哪一个输出发生变化。


## 示例 5：先切片，再用条件筛选

看懂上一个例子后，再观察同一主题在另一种数据或场景中的写法。


```python
import numpy as np

matrix = np.arange(1, 13).reshape(3, 4)
print("最后两行：\n", matrix[-2:])
print("第二列：", matrix[:, 1])
print("大于6的元素：", matrix[matrix > 6])
```

### 逐步拆解

切片通常保留数组形状；布尔筛选会返回满足条件的元素，适合快速定位重点值。

## 教学实验：axis与布尔筛选

这一组实验专门训练“观察一个结果 → 只改一个变量 → 解释变化”。先运行第一个代码单元格，再运行第二个。


```python
import numpy as np

matrix = np.arange(1, 13).reshape(3, 4)
print("原数组：\n", matrix)
print("每行合计：", matrix.sum(axis=1))
print("每列合计：", matrix.sum(axis=0))
```

### 第一个结果怎么读

`axis=1` 保留行，沿列方向计算；`axis=0` 保留列，沿行方向计算。先看 shape，再解释结果长度。

请记录：输入是什么、输出是什么、输出支持了哪一个结论。


```python
even = matrix[matrix % 2 == 0]
print("偶数：", even)
print("偶数数量：", even.size)
print("偶数平均值：", even.mean())
```

### 第二个结果怎么读

第二个实验不改原数组，而是用布尔条件筛选新数组。请思考：如果条件改成 `matrix > 8`，输出会怎样变化？

## 错误恢复：数组形状不匹配怎么办

真实数据和真实代码都会出问题。本节先观察问题，再用一个明确的检查或修复步骤恢复运行。


```python
import numpy as np

matrix = np.arange(6).reshape(2, 3)
try:
    result = matrix + np.array([10, 20])
except ValueError as error:
    print("形状问题：", type(error).__name__)
    result = matrix + np.array([10, 20, 30])
print("修复后的结果：")
print(result)
```
