## 核心概念

- 字符串是不可变序列，方法通常返回新字符串。

- 清洗前应保留原始值，并明确大小写、空格、分隔符和缺失值规则。


## 字符串方法分类速查

| 类别 | 常用方法 | 主要用途 | 需要特别注意 |
| --- | --- | --- | --- |
| 大小写 | `upper()`、`lower()`、`casefold()`、`capitalize()`、`title()`、`swapcase()` | 统一展示或比较格式 | 方法返回新字符串，不会修改原变量 |
| 空白处理 | `strip()`、`lstrip()`、`rstrip()` | 去除两端空白或指定字符 | 默认只处理两端，不会删除中间空格 |
| 查找统计 | `find()`、`rfind()`、`index()`、`rindex()`、`count()` | 定位子串和统计出现次数 | `find()` 找不到返回 -1，`index()` 找不到会报 `ValueError` |
| 前后缀 | `startswith()`、`endswith()`、`removeprefix()`、`removesuffix()` | 检查或删除固定前后缀 | 只处理开头或结尾，不是任意位置替换 |
| 替换 | `replace(old, new, count)` | 替换文本，可限制次数 | 不传 `count` 时替换所有匹配项 |
| 拆分组合 | `split()`、`rsplit()`、`splitlines()`、`partition()`、`join()` | 在字符串和列表之间转换 | `join()` 的元素必须是字符串 |
| 内容判断 | `isalpha()`、`isdigit()`、`isnumeric()`、`isalnum()`、`isspace()`、`islower()`、`isupper()` | 校验输入内容 | 空字符串通常返回 `False` |
| 对齐填充 | `center()`、`ljust()`、`rjust()`、`zfill()` | 生成固定宽度文本 | 宽度小于原长度时不会截断 |
| 格式化 | f-string、`format()`、格式说明符 | 组合变量、控制小数和宽度 | 格式化只是展示，不会改变原数据类型 |


## 示例 1：索引与切片

切片左闭右开，可以省略起止位置。


```python
text = "Python Data Analysis"
print(text[0], text[-1])
print(text[:6])
print(text[7:11])
print(text[::-1])
```

## 示例 2：大小写与空白处理

先去掉两端空白，再根据展示或比较需求选择大小写方法。


```python
raw = "  Python Data Analysis  "
print("原始文本:", repr(raw))
print("upper:", raw.upper())
print("lower:", raw.lower())
print("casefold:", raw.casefold())
print("capitalize:", raw.strip().capitalize())
print("swapcase:", raw.swapcase())
print("title:", raw.strip().title())
print("strip:", repr(raw.strip()))
print("lstrip:", repr(raw.lstrip()))
print("rstrip:", repr(raw.rstrip()))
```

## 示例 3：查找、统计与前后缀

find适合安全查找，index适合确认必须存在的字段；startswith和endswith用于判断边界。


```python
text = "report_2026_08.csv"
print("report位置:", text.find("report"))
print("最后一个下划线:", text.rfind("_"))
print("csv出现次数:", text.count("csv"))
print("是否以report开头:", text.startswith("report"))
print("是否以.csv结尾:", text.endswith(".csv"))
print("扩展名位置:", text.index("."))
print("最后一个点号:", text.rindex("."))
print("没有的内容:", text.find("xlsx"))
```

## 示例 4：替换与删除前后缀

replace可以限制替换次数；removeprefix和removesuffix只处理固定的开头或结尾。为了兼容较旧的Python版本，示例提供等价写法。


```python
raw = "待处理-订单-订单.csv"
print("替换全部:", raw.replace("订单", "销售"))
print("只替换一次:", raw.replace("订单", "销售", 1))
filename = "backup_report.csv"
if hasattr(filename, "removeprefix"):
    without_prefix = filename.removeprefix("backup_")
else:
    without_prefix = filename.replace("backup_", "", 1)
if hasattr(filename, "removesuffix"):
    without_suffix = filename.removesuffix(".csv")
else:
    without_suffix = filename[:-4] if filename.endswith(".csv") else filename
print("删除前缀:", without_prefix)
print("删除后缀:", without_suffix)
```

## 示例 5：拆分、反向拆分与partition

split适合拆成多个字段，rsplit可以从右侧开始，partition始终返回三部分。


```python
record = "A102|华东|2026|1280.50"
print("全部拆分:", record.split("|"))
print("从右侧拆一次:", record.rsplit("|", 1))
print("只拆第一处:", record.partition("|"))
print("只从右侧拆一处:", record.rpartition("|"))
print("逐行拆分:", "第一行\n第二行\n第三行".splitlines())
```

## 示例 6：内容判断方法

is开头的方法返回布尔值，适合在转换或清洗前检查输入。


```python
samples = ["Python", "2026", "Python2026", "²", "   ", "", "中文"]
for value in samples:
    print(repr(value), {
        "isalpha": value.isalpha(),
        "isdigit": value.isdigit(),
        "isnumeric": value.isnumeric(),
        "isalnum": value.isalnum(),
        "isspace": value.isspace(),
        "islower": value.islower(),
        "isupper": value.isupper(),
    })
```

## 示例 7：对齐、填充与格式化

对齐方法适合生成展示文本；f-string适合把变量和格式说明组合起来。


```python
code = "A7"
amount = 1280.5
print("居中:", code.center(8, "-"))
print("左对齐:", code.ljust(8, "."))
print("右对齐:", code.rjust(8, "."))
print("数字补零:", code.zfill(5))
print(f"金额：{amount:,.2f} 元")
print("编号：{}，金额：{:,.2f} 元".format(code, amount))
```

## 示例 8：拆分、清洗与重新组合

把复杂清洗拆成多个中间步骤，更容易发现是哪一步改变了数据。


```python
record = "  A102 | EAST | 1,280.50  "
parts = [part.strip() for part in record.split("|")]
order_id = parts[0].upper()
region = parts[1].casefold()
amount = float(parts[2].replace(",", ""))
clean_record = "|".join([order_id, region, f"{amount:.2f}"])
print("字段:", parts)
print("清洗后:", clean_record)
```

## 教学实验：边界条件

这一组实验专门训练“观察一个结果 → 只改一个变量 → 解释变化”。先运行第一个代码单元格，再运行第二个。


```python
orders = [
    {"order_id": "A01", "amount": 280},
    {"order_id": "A02", "amount": 300},
    {"order_id": "A03", "amount": 520},
]
threshold = 300
for order in orders:
    label = "达到门槛" if order["amount"] >= threshold else "未达到门槛"
    print(order["order_id"], order["amount"], label)
```

### 第一个结果怎么读

这里的重点不是记住 `if`，而是观察 `>=` 如何处理刚好等于 300 的记录。边界条件必须和业务规则保持一致。

请记录：输入是什么、输出是什么、输出支持了哪一个结论。


```python
threshold = 500
for order in orders:
    label = "重点关注" if order["amount"] >= threshold else "普通订单"
    print(order["order_id"], "->", label)
print("重点订单数：", sum(order["amount"] >= threshold for order in orders))
```

### 第二个结果怎么读

只把门槛从 300 改成 500，再比较标签和数量变化。这个实验训练的是“改一个输入，解释一个输出”。

## 错误恢复：类型转换失败怎么办

真实数据和真实代码都会出问题。本节先观察问题，再用一个明确的检查或修复步骤恢复运行。


```python
amount_text = "128.5"
try:
    amount = int(amount_text)
except ValueError as error:
    print("第一次转换失败：", type(error).__name__)
    amount = float(amount_text)
print("可以继续使用的金额：", amount)
```

### 关键知识速查

| 知识点 | 作用与提醒 | 关键写法 |
| --- | --- | --- |
| 索引与切片 | 切片左闭右开，可以省略起止位置。 | `text[0]`、`text[-1]`、`text[:6]`、`text[7:11]` |
| 大小写与空白处理 | 先去掉两端空白，再根据展示或比较需求选择大小写方法。 | `raw.upper()`、`raw.lower()`、`raw.casefold()`、`raw.strip()` |
| 查找、统计与前后缀 | find适合安全查找，index适合确认必须存在的字段；startswith和endswith用于判断边界。 | `text.find()`、`text.rfind()`、`text.count()`、`text.startswith()` |
| 替换与删除前后缀 | replace可以限制替换次数；removeprefix和removesuffix只处理固定的开头或结尾。为了兼容较旧的Python版本，示例提供等价写法。 | `raw.replace()`、`filename.removeprefix()`、`filename.replace()`、`filename.removesuffix()` |
| 拆分、反向拆分与partition | split适合拆成多个字段，rsplit可以从右侧开始，partition始终返回三部分。 | `record.split()`、`record.rsplit()`、`record.partition()`、`record.rpartition()` |
| 内容判断方法 | is开头的方法返回布尔值，适合在转换或清洗前检查输入。 | `value.isalpha()`、`value.isdigit()`、`value.isnumeric()`、`value.isalnum()` |
| 对齐、填充与格式化 | 对齐方法适合生成展示文本；f-string适合把变量和格式说明组合起来。 | `code.center()`、`code.ljust()`、`code.rjust()`、`code.zfill()` |
| 拆分、清洗与重新组合 | 把复杂清洗拆成多个中间步骤，更容易发现是哪一步改变了数据。 | `part.strip()`、`record.split()`、`.upper()`、`.casefold()` |


### 需要注意

- 忘记字符串方法不会原地修改原变量
- 把strip误认为会删除中间空格
- 混淆find返回-1和index抛出ValueError
- replace默认替换所有匹配项却没有检查数量
- 把isdigit、isnumeric和isalpha混为一谈
- split后的字段数量与解包变量数量不一致
- join的列表中混入数字而不是字符串
