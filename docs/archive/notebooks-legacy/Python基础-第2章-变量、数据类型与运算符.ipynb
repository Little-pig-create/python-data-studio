## 核心概念

- 变量名应表达含义，并避免覆盖Python内置函数。
- 除法通常得到float；金额、比例和数量要明确单位。
- 比较运算产生bool，多个条件使用and、or和not组合。


## 示例 1：变量与类型

type可以确认运行时的数据类型。


```python
product = "键盘"
quantity = 3
unit_price = 299.0
is_member = True
coupon = None
for value in [product, quantity, unit_price, is_member, coupon]:
    print(repr(value), type(value).__name__)
```

## 示例 2：指标计算

先给中间指标命名，避免把复杂公式压成一行。


```python
subtotal = quantity * unit_price
discount_rate = 0.9 if is_member else 1.0
payable = subtotal * discount_rate
print(f"原价: {subtotal:.2f} 元")
print(f"应付: {payable:.2f} 元")
```

## 示例 3：比较与逻辑运算

布尔表达式可以直接描述规则。


```python
free_shipping = payable >= 99
needs_review = quantity >= 10 or payable >= 5000
valid_order = quantity > 0 and unit_price > 0
print("包邮:", free_shipping)
print("需要复核:", needs_review)
print("订单有效:", valid_order)
```

## 示例 4：把金额计算拆成中间变量

这一组例子只处理一个小问题。先运行代码，再逐行对照拆解说明。


```python
unit_price = 199.0
quantity = 2
discount_rate = 0.90
subtotal = unit_price * quantity
discount = subtotal * (1 - discount_rate)
payable = subtotal - discount
print(f"小计：{subtotal:.2f} 元")
print(f"优惠：{discount:.2f} 元")
print(f"应付：{payable:.2f} 元")
```

### 逐步拆解

先算小计，再算优惠，最后得到应付金额。每个变量只承担一个含义，便于检查公式。

建议第一次运行后只改一个输入值，再观察哪一个输出发生变化。


## 示例 5：单位转换与类型检查

看懂上一个例子后，再观察同一主题在另一种数据或场景中的写法。


```python
temperature_c = 26
temperature_f = temperature_c * 9 / 5 + 32
print(f"{temperature_c}℃ = {temperature_f:.1f}℉")
print("摄氏温度类型：", type(temperature_c).__name__)
print("华氏温度类型：", type(temperature_f).__name__)
```

### 逐步拆解

除法会产生浮点数；用 type(...).__name__ 可以快速确认变量的运行时类型。

## 教学实验：边界条件

这一组实验专门训练“观察一个结果 → 只改一个变量 → 解释变化”。先运行第一个代码单元格，再运行第二个。


```python
orders = [
    {"order_id": "A01", "amount": 280},
    {"order_id": "A02", "amount": 300},
    {"order_id": "A03", "amount": 520},
]
threshold = 300
for order in orders:
    label = "达到门槛" if order["amount"] >= threshold else "未达到门槛"
    print(order["order_id"], order["amount"], label)
```

### 第一个结果怎么读

这里的重点不是记住 `if`，而是观察 `>=` 如何处理刚好等于 300 的记录。边界条件必须和业务规则保持一致。

请记录：输入是什么、输出是什么、输出支持了哪一个结论。


```python
threshold = 500
for order in orders:
    label = "重点关注" if order["amount"] >= threshold else "普通订单"
    print(order["order_id"], "->", label)
print("重点订单数：", sum(order["amount"] >= threshold for order in orders))
```

### 第二个结果怎么读

只把门槛从 300 改成 500，再比较标签和数量变化。这个实验训练的是“改一个输入，解释一个输出”。

## 错误恢复：类型转换失败怎么办

真实数据和真实代码都会出问题。本节先观察问题，再用一个明确的检查或修复步骤恢复运行。


```python
amount_text = "128.5"
try:
    amount = int(amount_text)
except ValueError as error:
    print("第一次转换失败：", type(error).__name__)
    amount = float(amount_text)
print("可以继续使用的金额：", amount)
```

### 错误恢复步骤

1. 先看错误类型、字段或数据形状。
2. 判断问题发生在输入、处理中间结果还是输出。
3. 修复后重新检查结果，而不是只让代码不报错。

先读错误类型，再决定修复方法。这里不是盲目忽略错误，而是明确知道整数转换不适合带小数的文本。
